# Flashier - Timing (3 runs)

💡 **Environment:** `clamp-analyses`  

```r
library(here)
library(flashier)
library(ebnm)
set.seed(3)
```

```r
output_dir <- here("output/model_performance/gtex")
dir.create(output_dir, showWarnings = FALSE, recursive = TRUE)

N_RUNS <- 3
```

```r
gtex_data <- readRDS(here("output/gtex/df_gtex_fbm_filt.rds"))
K <- readRDS(here("output/gtex/CLAMP_K_gtex.rds"))
K <- as.integer(K)

X <- t(as.matrix(gtex_data))
```

```r
flashier_times <- numeric(N_RUNS)

for (i in 1:N_RUNS) {
  cat("Flashier run", i, "of", N_RUNS, "\n")
  
  start_time <- Sys.time()
  
  fl <- flash_init(X, var_type = 0L)
  
  fl <- flash_greedy(
    fl,
    Kmax = as.integer(K),su
    ebnm_fn = c(ebnm_point_exponential, ebnm_point_normal),
    verbose = 1L
  )
  
  fl <- flash_backfit(fl, verbose = 1L, maxiter = 20)
  
  end_time <- Sys.time()
  flashier_times[i] <- as.numeric(difftime(end_time, start_time, units = "mins"))
  cat("Run", i, "time:", flashier_times[i], "minutes\n\n")
}
```

```r
flashier_time_minutes <- flashier_times
names(flashier_time_minutes) <- paste0("run", 1:N_RUNS)
saveRDS(flashier_time_minutes, file.path(output_dir, "flashier_time_minutes.rds"))
cat("Flashier times:", flashier_time_minutes, "minutes\n")
```
