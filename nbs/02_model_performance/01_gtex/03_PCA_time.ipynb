# PCA - Timing (3 runs)

```python
from pathlib import Path
import time
import numpy as np
import pandas as pd
from pyprojroot.here import here
from sklearn.decomposition import PCA
import pickle
```

```python
K = pd.read_csv(here("output/gtex/CLAMP_K_gtex.csv"), index_col=0)
n_components = int(K["CLAMP_K_gtex"].iloc[0])
display(print(n_components))

output_dir = Path(here("output/model_performance/gtex"))
output_dir.mkdir(parents=True, exist_ok=True)

gtex_data = pd.read_csv(here("output/gtex/df_gtex_fbm_filt.csv"), index_col=0).astype(np.float32)
N_RUNS = 3
```

```python
X = gtex_data.T  # samples x genes

PCA_times = []

for i in range(N_RUNS):
    print(f"PCA run {i+1} of {N_RUNS}")
    
    start_time = time.time()
    
    pca = PCA(n_components=n_components, svd_solver="auto", random_state=0)
    W = pca.fit_transform(X)
    
    end_time = time.time()
    elapsed = (end_time - start_time) / 60
    PCA_times.append(elapsed)
    print(f"Run {i+1} time: {elapsed} minutes\n")
```

```python
PCA_time_minutes = {f"run{i+1}": t for i, t in enumerate(PCA_times)}
with open(output_dir / "PCA_time_minutes.pkl", "wb") as f:
    pickle.dump(PCA_time_minutes, f)
print(f"PCA times: {PCA_time_minutes}")
```
