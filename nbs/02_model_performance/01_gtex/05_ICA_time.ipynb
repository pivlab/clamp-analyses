# ICA - Timing (3 runs)

💡 **Environment:** `clamp-analyses`  

```python
from pathlib import Path
import time
import numpy as np
import pandas as pd
from pyprojroot.here import here
from sklearn.decomposition import FastICA
import pickle
```

```python
K = pd.read_csv(here("output/gtex/CLAMP_K_gtex.csv"), index_col=0)
n_components = int(K["CLAMP_K_gtex"].iloc[0])
display(print(n_components))

output_dir = Path(here("output/model_performance/gtex"))
output_dir.mkdir(parents=True, exist_ok=True)

gtex_data = pd.read_csv(here("output/gtex/df_gtex_fbm_filt.csv"), index_col=0).astype(np.float32)
N_RUNS = 3
```

```python
X = gtex_data.T  # samples x genes

ICA_times = []

for i in range(N_RUNS):
    print(f"ICA run {i+1} of {N_RUNS}")
    
    start_time = time.time()
    
    ica = FastICA(n_components=n_components, random_state=0, max_iter=2000)
    W = ica.fit_transform(X)
    
    end_time = time.time()
    elapsed = (end_time - start_time) / 60
    ICA_times.append(elapsed)
    print(f"Run {i+1} time: {elapsed} minutes\n")
```

```python
ICA_time_minutes = {f"run{i+1}": t for i, t in enumerate(ICA_times)}
with open(output_dir / "ICA_time_minutes.pkl", "wb") as f:
    pickle.dump(ICA_time_minutes, f)
print(f"ICA times: {ICA_time_minutes}")
```
