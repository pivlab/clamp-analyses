```python
import rpy2.robjects as ro
from rpy2.robjects.conversion import localconverter
from rpy2.robjects import pandas2ri
readRDS = ro.r["readRDS"]

from pyprojroot.here import here

gtex_clamp_full = readRDS(str(here("output/gtex/CLAMPfull.rds")))
gtex_genes = readRDS(str(here("output/gtex/gtex_genes.rds")))
gtex_samples = readRDS(str(here("output/gtex/gtex_samples.rds")))

summary_matrix = gtex_clamp_full.rx2("summary")

with localconverter(ro.default_converter + pandas2ri.converter):
    summary_values = ro.conversion.rpy2py(summary_matrix)

gtex_summary = pd.DataFrame(
    data=summary_values,
    index=summary_matrix.rownames if summary_matrix.rownames else None,
    columns=summary_matrix.colnames if summary_matrix.colnames else None
)
gtex_summary.head()

Z_matrix = gtex_clamp_full.rx2("Z")

with localconverter(ro.default_converter + pandas2ri.converter):
    Z_values = ro.conversion.rpy2py(Z_matrix)
    genes = ro.conversion.rpy2py(gtex_genes)

colnames = ro.r['colnames'](Z_matrix)

gtex_Z = pd.DataFrame(
    data=Z_values,
    index=list(genes),
    columns=list(colnames)
)
gtex_Z.head()



B_matrix = gtex_clamp_full.rx2("B")

with localconverter(ro.default_converter + pandas2ri.converter):
    B_values = ro.conversion.rpy2py(B_matrix)
    samples = ro.conversion.rpy2py(gtex_samples)

rownames = ro.r['rownames'](B_matrix)

gtex_B = pd.DataFrame(
    data=B_values,
    index=list(rownames),
    columns=list(samples)
)

gtex_B.head()
```
