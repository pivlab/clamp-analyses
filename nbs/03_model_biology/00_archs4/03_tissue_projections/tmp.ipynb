```r
library(here)
```

```r
archs4 <- readRDS(here('output/01_model_building/04_archs4/06_bp_coverage_rshall/06_bp_coverage_hall_rs_100/CLAMPfull_hall.rds'))
```

```r
gtex <- readRDS(here('output/01_model_building/02_gtex/10_CLAMP_hall/CLAMPfull_hall.rds'))
```

```r
archs4_summary <- data.frame(as.matrix(archs4$summary))
```

```r
head(archs4_summary)
```

```r
unique(archs4_summary$pathway)
```

```r
gtex_summary <- data.frame(as.matrix(gtex$summary))
```

```r
head(gtex_summary)
```

```r
unique(gtex_summary$pathway)
```

```r
library(dplyr)

archs4_sig <- archs4_summary %>%
  mutate(
    AUC = as.numeric(AUC),
    FDR = as.numeric(FDR)
  ) %>%
  filter(FDR < 0.05, AUC > 0.7) %>%
  distinct(pathway)

gtex_sig <- gtex_summary %>%
  mutate(
    AUC = as.numeric(AUC),
    FDR = as.numeric(FDR)
  ) %>%
  filter(FDR < 0.05, AUC > 0.7) %>%
  distinct(pathway)

archs4_only_sig_pathways <- archs4_sig %>%
  anti_join(gtex_sig, by = "pathway")
```
