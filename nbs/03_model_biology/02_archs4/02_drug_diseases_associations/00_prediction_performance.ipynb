# Canonical drug-disease performance

💡 **Environment:** `clamp-analyses`

Drug repurposing by CLAMP LV-space scoring (PhenoPlier-style): project LINCS L1000
drug signatures and S-PrediXcan gene-trait z-scores into each compendium's canonical
CLAMPfull model, then score drug-disease pairs as `-drug^T . disease` in LV space.
This reports the headline AUROC per compendium (ARCHS4, recount2, GTEx),
mean-over-tissue-threshold then max-over-tissue aggregated across all 49 GTEx
tissues against the PhenomeXcan gold standard.

```python
from pathlib import Path
import pandas as pd
from IPython.display import display

ROOT = Path(snakemake.input.aggregate)
summary = pd.read_csv(ROOT / 'performance_summary.csv', index_col=0)
assert set(summary.index) == {'module_based_archs4', 'module_based_recount2', 'module_based_gtex', 'gene_based'}
rename = {'module_based_archs4':'ARCHS4', 'module_based_recount2':'recount2', 'module_based_gtex':'GTEx', 'gene_based':'Gene-based'}
summary = summary.rename(index=rename)[['AUROC']]
summary.loc['Random (chance)'] = [0.5]
display(summary)
```
